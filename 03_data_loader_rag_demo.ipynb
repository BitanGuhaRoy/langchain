# Data Loader

```python
from langchain_community.document_loaders import PyPDFLoader
PdfTextLoader = PyPDFLoader("rag_data/rag_knowledge_dataset.pdf")
documents= PdfTextLoader.load()
print(f"Number of documents loaded: {len(documents)}")
```

```python
from langchain_community.document_loaders import UnstructuredHTMLLoader
HtmlTextLoader = UnstructuredHTMLLoader("./rag_data/rag_knowledge_dataset.html")
html_documents = HtmlTextLoader.load()
print(f"Number of HTML documents loaded: {len(html_documents)}")
```
