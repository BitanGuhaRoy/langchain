```python
from dotenv import load_dotenv

load_dotenv()

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma

```

```python
# load documents 
PdfTextLoader = PyPDFLoader("rag_data/rag_knowledge_dataset.pdf")
documents= PdfTextLoader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=0)
docs = splitter.split_documents(documents)
print(len(docs))
```

```python
#create embeddings 
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)
```

```python
vectorstore = Chroma.from_documents(docs, embedding=embeddings, persist_directory="./chroma_db")
```

```python
question = "What is message queue ?"

```

```python
results = vectorstore.similarity_search(question, k=3)
for i, result in enumerate(results):
    print(f"Result {i+1}:")
    print(result.page_content)
    print("\n---\n")
```

```python
from langchain_openai import ChatOpenAI
question = "What is load balancer  ?"


llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
context = "\n\n".join([result.page_content for result in results])

# Create prompt
prompt = f"""
You are a company Software architect.

Answer the question using ONLY the provided context.

If the answer is not present in the context,
say that you don't know.

Context:
{context}

Question:
{question}
"""


response = llm.invoke(prompt)
print("Response from LLM:")
print(response.content)
```

```python
retriever = vectorstore.as_retriever()
response = retriever.invoke("what  is queue?")
print(retriever)
```

```python
# Simple use with LCEL 
from langchain_core.runnables import RunnablePassthrough
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
template = """
You are a company Software architect.
Answer the question using ONLY the provided context.
If the answer is not present in the context,
say that you don't know.
Context:
{context}
Question:
{question}
"""
```

```python
prompt =ChatPromptTemplate.from_template(template)
model = ChatOpenAI()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])

chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()} | prompt | model | StrOutputParser()
)
```

```python
response = chain.invoke("What is the leave policy?")

print(response)
```

```python
question = "Who is Biden ?"

docs = retriever.invoke(question)

context = format_docs(docs)

messages = prompt.invoke({
    "context": context,
    "question": question
})

response = model.invoke(messages)
print(response.content)
```
